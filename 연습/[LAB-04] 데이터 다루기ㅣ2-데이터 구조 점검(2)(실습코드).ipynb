## #02. 행,열 재배치

### 1. 열 순서 변경

### 2. 행 순서 변경

- 지정된 순서대로 행이 재정렬 된 결과가 반환된다.
- 원본은 변화 없음. 결과가 적용된 복사본이 생성된다.
- 재배치 되지 못한 행은 제외된다.

```python
행순서변경DF = origin.reindex(index=['석영', '호영', '민철', '수현'])
행순서변경DF
```

## #03. 행 추가하기

### 1. 리스트로 행 추가하기

- 추가될 행의 인덱스 이름을 지정해야 한다.
- 리스트로 추가할 경우 DataFrame의 컬럼 순서에 맞게 지정해야 한다.
- 누락되는 값이 있거나 값의 수가 초과할 경우 에러

```python
리스트행추가DF = origin.copy()
리스트행추가DF.loc['정호'] = [2, '남자', 90, 80, 70, 60]
리스트행추가DF
```

### 2. 딕셔너리로 추가하기

-   컬럼 순서는 상관 없다.
-   누락되는 값이 있거나 값의 수가 초과할 경우 에러

```python
딕셔너리행추가DF = origin.copy()
딕셔너리행추가DF.loc['민정'] = {"학년": 4, "성별": "여자", "국어": 81, "영어": 72, "수학": 84, "과학": 90}
딕셔너리행추가DF
```

### 3. Series 객체로 행 추가하기

- 순서는 상관 없다.
- Series 객체의 index가 DataFrame의 변수명과 동일해야 한다.

```python
시리즈행추가DF = origin.copy()
s = Series([2, "남자", 90, 80, 70, 60],
           index=["학년", "성별", "국어", "영어", "수학", "과학"])
시리즈행추가DF.loc['형석'] = s
시리즈행추가DF
```

### 4. 빈 값을 포함하는 행 추가하기

- 데이터를 추가할 때 값이 없는 필드는 생략할 수 있다.
- Series 형태의 객체로 추가할 때만 가능함

```python
빈값포함추가DF = origin.copy()
d = Series([2, "남자", 93, 71],
           index=["학년", "성별", "국어", "수학"])
빈값포함추가DF.loc['석호'] = d
빈값포함추가DF
```

### 5. 기존의 행을 복사해서 행 추가 하기

```python
기존행복사DF = origin.copy()
기존행복사DF.loc['철민'] = 기존행복사DF.loc['철수']
기존행복사DF
```

### 6. 단일 값으로 행 추가하기

- 모든 변수에 같은 값이 저장된다.

```python
단일값추가DF = origin.copy()
단일값추가DF.loc['희영'] = 100
단일값추가DF
```

## #04. 열 추가하기

### 1. 리스트로 열 추가하기

- 모든 행에 대한 값을 순서대로 나열한다.
- 기존의 행과 비교하여 추가되는 값이 부족하거나 초과하면 에러 발생
- 빈값을 추가하고자 하는 경우 해당 위치에 `None`이라고 명시

```python
리스트열추가DF = origin.copy()
리스트열추가DF['한국사'] = [72, 56, 90, None,92, 90, 75, 80, 60]
리스트열추가DF
```

### 2. 딕셔너리로 열 추가하기

- 추가되지 않을 행은 명시하지 않는다.
- 데이터프레임에 존재하지 않는 인덱스가 포함될 경우 해당 값을 버린다.
- 명시되지 않은 행은 빈값(`NaN`)으로 설정된다.

```python
딕셔너리열추가DF = origin.copy()
딕셔너리열추가DF['세계사'] = {"철수": 90, "영희": 88, "민철":72, "수현":56, "정우":90, "준명": 100, "용식":77, "나영":95, "석영":54}
딕셔너리열추가DF
```

### 3. Series 객체로 추가하기

- 각 값이 연결될 행의 이름(index)를 지정해야 한다.
- 부분적으로 값을 비워 둘 수 있다.

```python
시리즈열추가DF = origin.copy()
s = Series([72, 56, 90, 88], index=['철수', '민철', '호영', '용식'])
시리즈열추가DF['윤리'] = s
시리즈열추가DF
```

### 4. 기존의 열을 복사해서 열 추가하기

```python
열복사DF = origin.copy()
열복사DF['체육'] = 열복사DF['과학']
열복사DF
```

### 5. 단일 값으로 열 추가하기

```python
단일값열추가DF = origin.copy()
단일값열추가DF['체육'] = 100
단일값열추가DF
```

## #05. 행 삭제하기

### 1. 단일 행 삭제

- 추가는 원본 자체가 변하지만 삭제는 결과가 반영된 새로운 복사본이 생성된다.
- `drop()` 메서드도 `inplace=True` 파라미터 적용 가능


```python
단일행삭제DF = origin.copy()
단일행삭제DF = 단일행삭제DF.drop(index="철수")
단일행삭제DF
```

- 삭제할 행의 인덱스를 리스트로 지정한다.

```python
다중행삭제DF = origin.copy()
다중행삭제DF = 다중행삭제DF.drop(index=['철수', '민철', '수현'])
다중행삭제DF
```

### 3. index에 대한 인덱싱으로 행 삭제

```python
인덱싱행삭제철수DF = origin.copy()
인덱싱행삭제철수DF = 인덱싱행삭제철수DF.drop(index=인덱싱행삭제철수DF.index[0])
인덱싱행삭제철수DF
```

### 4. index에 대한 슬라이싱으로 행 삭제

```python
인덱싱행삭제슬DF = origin.copy()
인덱싱행삭제슬DF = 인덱싱행삭제슬DF.drop(index=인덱싱행삭제슬DF.index[0:7])
인덱싱행삭제슬DF
```

## #06 열 삭제

### 1. 단일 열 삭제

- 행 삭제는 `drop()` 메서드에 `index` 파라미터를 설정하고,     
  열 삭제는 `columns` 파라미터를 설정한다.

```python
단일열삭제DF = origin.copy()
단일열삭제DF = 단일열삭제DF.drop(columns='학년')
단일열삭제DF
```

### 2. 다중 열 삭제

- `columns` 파라미터에 삭제 대상 변수 이름을 리스트 타입으로 설정한다.

```python
다중열삭제DF = origin.copy()
다중열삭제DF = 다중열삭제DF.drop(columns=['학년', '성별', '과학'])
다중열삭제DF
```

### 3. 컬럼명에 대한 인덱싱으로 열 삭제

```python
인덱싱열삭제DF = origin.copy()
인덱싱열삭제DF = 인덱싱열삭제DF.drop(columns=인덱싱열삭제DF.columns[1])
인덱싱열삭제DF
```

### 4. 컬럼명에 대한 슬라이싱으로 열 삭제

```python
인덱싱열삭제슬DF = origin.copy()
인덱싱열삭제슬DF = 인덱싱열삭제슬DF.drop(columns=인덱싱열삭제슬DF.columns[2:4])
인덱싱열삭제슬DF
```
