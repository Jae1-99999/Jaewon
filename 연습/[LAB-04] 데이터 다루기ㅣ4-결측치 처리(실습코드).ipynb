## #03. 결측치 소거

### 1. 행 단위 삭제

결측치가 있는 모든 행 삭제 (원본은 변화 없음, 삭제 결과가 리턴됨)

`inplace=True` 파라미터를 적용할 경우 원본에 즉시 반영되고 리턴값 없음

### 2. 열 단위 삭제

### 3. 특정 열에 대해서만 결측치가 존재하는 행 삭제

```python
#kor열에 존재하는 행 삭제
na3 = origin.dropna(subset=['kor'])
na3
```

## #04. 결측치 대체

### 1. 고정값으로 대체

모든 결측치를 동일한 값으로 교체

원본은 변화 없음, 삭제결과 리턴됨

`inplace=True`를 적용할 경우 원본에 즉시 반영되고 리턴값 없음

```python
re_df1 = origin.fillna(value=50)
re_df1
```

### 2. 통계적 값으로 대체 (1) - 결측치를 정제할 규칙을 담고 있는 객체 생성

각 열단위로 평균`(strategy='mean')`을 결측치`(missing_values)`에 지정

strategy 옵션 : mean=평균, median=중앙값, most_frequent: 최빈값(가장 많이 관측되는 수)

> 이 객체의 생성이 결측치를 즉시 처리했다는 것을 의미하는 것은 아니다.

```python
imr = SimpleImputer(missing_values=np.nan, strategy='mean')
```

### 3. 통계적 값으로 대체 (2) - 생성된 규칙을 적용

결측치가 정제된 2차원 배열로 리턴되므로 다시 DataFrame으로 재구성할 필요가 있다.

```python
df_imr = imr.fit_transform(origin.values)
df_imr
```

### 4. 통계적 값으로 대체 (3) - 결측치 정제 결과를 DataFrame으로 재구성

```python
re_df2 = DataFrame(df_imr,
                   index=origin.index,
                   columns=origin.columns
                   )
re_df2
```

### 5. 특정 컬럼 하나만 처리하기 (1)- 컬럼의 값만 추출한 후 n행1열로 재구성

- sklearn의 SimpleInputer는 파라미터로 2차 배열을 요구한다.
- 특정 컬럼 하나에 대해서만 결측치 대체를 수행해야 할 경우    
  sklearn의 요구사항에 맞춰 n행 1열로 재구성해야 한다.

```python
reshape = origin['kor'].values.reshape(-1,1)
reshape
```

### 6. 특정 컬럼의 값 하나만 처리하기 (2) - 결측치 처리하기

```python
reshape_imr = imr.fit_transform(reshape)
reshape_imr
```

### 7. 특정 컬럼의 값 하나만 처리하기 (3) - 결측치 처리 결과를 DataFrame의 컬럼에 적용하기

```python
특정컬럼_정제df = origin.copy()
특정컬럼_정제df['kor'] = reshape_imr
특정컬럼_정제df
```

```python
# 연습문제
# 안녕하세요, 데이터 분석가님
# 최근 한달간의 판매 기록(month_sales)을 받았는데, 아쉽게도 일부 데이터가 누락되었습니다.
# 정확한 월말 매출 보고를 위해 분석가님의 도움이 필요합니다.
# 아래의 미션을 순서대로 해결하여 데이터 정제를 완료.
# 미션1 : 데이터 탐색
#   데이터를 불러온 후 각 열(column)에 몇 개의 결측치 (missing value)가 있는지 확인해주세요.
# 미션2 : 신뢰할 수 없는 데이터 제거
#   'Quantity'(수량) 정보가 누락된 데이터는 주문 과정에서 오류가 있었을 가능성이 높습니다.
#   이 데이터는 분석에서 제외하기로 결정했습니다.
#    'Quantity'열에 결측치가 있는 모든 행(row)을 삭제해주세요.
# 미션3 : 가격 정보 복원
#   일부 제품의 'Price' (가격) 정보가 누락되었습니다.
#   해당 제품들은 'Price' 열의 전체 평균 가격으로 값을 채워 넣어 데이터를 복원해주세요.
```

```python
# 0. 파일 불러오기 및 정보 확인
MS = load_data('month_sales')
MS.info()
MS
# 미션 1. 데이터 탐색
MS_empty = MS.isnull()
MS_empty.sum()
# 미션2. 신뢰할 수 없는 데이터 제거
MS_Quantity = MS.dropna(subset=['Quantity'])
MS_Quantity
# 미션3. 가격 정보 복원
MS_reshape = MS_Quantity['Price'].values.reshape(-1,1)
imr = SimpleImputer(missing_values=np.nan, strategy='mean')
MS_imr = imr.fit_transform(MS_reshape)

MS_price = MS_Quantity.copy()
MS_price['Price'] = MS_imr
MS_price
```
