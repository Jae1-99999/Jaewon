## #02. 선 모양 설정 하기

### 1. 선 색상 / 선 스타일, 굵기 / 마커 모양, 크기, 테두리 색상, 면 색상, 테두리 굵기

```python
# 그래프 초기화 / 제목:그래프 선 모양 설정, x라벨 : 날짜, y라벨 : 판매수량
fig, ax = my_plot.init(title= "그래프 선 모양 설정", xlabel= "날짜", ylabel= "판매수량")

# 그래프 그리기 / x데이터 : Date, y데이터 : Quantity, color : #1D84C3, 선스타일 : 점선, 선굵기 : 5, 마커 : o, 마커크기 : 20, 마커굵기 : 3, 마커색깔 : #00ff00, 마커외곽색깔 : #ff0000
sb.lineplot(data=origin, x = "Date", y = "Quantity", c = "#1D84C3", linestyle = ":", linewidth = 5, marker="o", markersize=20 , markeredgewidth=3, markerfacecolor = "#00ff00", markeredgecolor="#ff0000")

# 출력
plt.show()
```

## #03. 축 설정

### 1. 축 범위/표시할 값 설정하기

- 축 범위 설정:
  - `ax`객체의 `set_xlim(), set_ylim()` 메서드를 사용한다.
- 축에 표시할 값 설정
  - `ax`객체의 `set_xticks(), set_yticks()` 메서드를 사용한다.

```python
# 그래프 초기화 / 제목 : y축 범위 설정, x라벨 : 날짜, y라벨 : 판매수량
fig, ax = my_plot.init(title = "y축 범위 설정", xlabel="날짜", ylabel="판매수량")

# 그래프 그리기 / x데이터 : Date, y데이터 : Quantity
sb.lineplot(data=origin, x="Date", y="Quantity")

# y축 범위 (x축 범위 설정은 set_xlim) / y축 범위 : 18~28
ax.set_ylim(18,28)

# y축에 표시할 값 설정
# --> [18~28]에 표시할 내용 내용을 [18~28] 범위의 실제 숫자로 지정
# --> 글자크기=6, 표시각도=-45도
ax.set_yticks(list(range(18,29)), list(range(18,29)), fontsize=6, rotation=-45)

# 출력
my_plot.show()
```

## #04. 다중 선 그래프

### 1. 두 개 이상의 선 그리기

- 여러 라인 플롯을 동시에 그리는 경우에는 각 선이 무슨 자료를 표시하는지를 보여주기 위해 `ax.legend()` 명령으로  범례(legend)를 추가할 수 있다.
- 범례의 위치는 자동으로 정해지지만 수동으로 설정하고 싶으면 `loc` 파라미터에 위치를 의미하는 문자열 혹은 숫자를 설정한다. 

```python
# 그래프 초기화 / 제목 : 범주별 그래프 그리기, X라벨 : X, Y라벨 : Y
fig, ax = my_plot.init(title="범주별 그래프 그리기", xlabel="X", ylabel="Y")

# 그래프 그리기 / A : x= 1,2,3,4 , y=10,20,35,40 라벨:A변화량 / B : x= 1,2,3,4 y= 10,15,30,35 라벨:B변화량
sb.lineplot(x=[1,2,3,4], y=[10,20,35,40], label="A변화량")
sb.lineplot(x=[1,2,3,4], y=[10,15,30,35], label="B변화량")
ax.legend(loc='upper left')   # 범주의 위치 설정 (미설정시 자동으로 위치가 지정됨) 범주 위치설정 : upper left

# 출력
plt.show()
```

#### 연습문제 Covid19확진자 수 변동 추이 시각화

```python
# covid19_active 데이터 셋 시각화
# 이 데이터 셋은 2022년 5월 1일부터 2023년 5월 31일까지 서울과 전국의 Covid19 일일 확진자 수를 기록한 데이터
# 조사 기간동안 서울과 전국의 확진자 수가 어떻게 변화하고 이쓴ㄴ지에 대한 추이를 하나의 그래프에 구분하여 시각화 하고 다음의 물음에 답하시오.
# hue파라미터를 사용하기 위해 melt함수를 활용한 데이터 재배치가 필요
```

```python
# 문제 1. 
# 조사기간 중 전국 일일 확진자 수가 가장 많았던 시점은 언제인가? (연도와 월)

# 0. 필요한 모듈 불러오기
import numpy as np
from jussam import load_data
from helper import my_plot
import seaborn as sb
from pandas import DataFrame, pivot_table, melt

# 1. 데이터 불러오기 및 정보 확인
covid19 = load_data('covid19_active')
covid19.info()
covid19
```

```python
# 2. hue파라미터를 사용하기 위한 melt함수를 활용한 데이터 재배치
covid19 = covid19.reset_index()
covid19

covid19_2 = melt(covid19, id_vars='기준일', value_vars=['서울시 일일 확진', '전국 일일 확진'],
                 var_name='구분', value_name='확진자수')
covid19_2
```

```python
# 3. 시각화
my_plot.lineplot(data=covid19_2, x='기준일', y='확진자수', hue='구분')
plt.show()
```

#### 문제1. 전국 일일확진자 수가 가장 많았던 시점
##### 주황색 선이 전국일일확진자 수인데 2022년 08월 시점에 확진자 수가 많이 늘어났다.

#### 문제2. 확진자의 수의 큰 봉우리(유행정점)는 총 몇번 나타났는가?
##### 전국에서 유행 시점은 2022년 08월 시점에 한번으로 유행정점시기가 있는데 서울시에서는 2022년 08월 시점에 한번 2022년 12월에 한번 총 두번의 유행정점이 있습니다.
#### 문제3. 서울시와 전국의 확진자 수 중, 항상 더 많은 쪽은 어디?
##### 그래프에 전국 통계가 주황색실선, 서울은 파란색 실선으로 전국일일확진자수가 항상 더 많았다.
